$$p_X(x) = \begin{cases} 0.50 & \text{if } x = 0 \\ 0.30 & \text{if } x = 1 \\ 0.15 & \text{if } x = 2 \\ 0.05 & \text{if } x = 3 \\ 0 & \text{otherwise} \end{cases}$$

Construct a Probability Space $(\Omega, \mathcal{F}, P)$Sample Space ($\Omega$): Since $X$ represents a proportion, the set of all possible outcomes is the unit interval.$$\Omega = [0, 1]$$$\sigma$-algebra ($\mathcal{F}$): We use the Borel $\sigma$-algebra on the interval $[0, 1]$, which includes all measurable subsets (intervals, points, etc.).$$\mathcal{F} = \mathcal{B}([0, 1])$$Probability Measure ($P$): The probability of any event $A \in \mathcal{F}$ is defined by the integral of the density function over that set.$$P(A) = \int_A f(x) \, dx$$

2. Define $X$ as a Continuous Random Variable$X$ is a function $X: \Omega \to [0, 1]$ that maps each outcome $\omega \in \Omega$ to itself:$$X(\omega) = \omega, \quad \forall \omega \in [0, 1]$$Because the probability of $X$ taking any single point value is zero ($P(X=x) = 0$), and it is defined over a continuum, it is a continuous random variable.

3. Probability Density Function (PDF)The PDF for a Beta distribution with parameters $\alpha = 2$ and $\beta = 18$ is:$$f(x) = \begin{cases} \frac{1}{B(2,18)} x^1 (1-x)^{17} & 0 < x < 1 \\ 0 & \text{otherwise} \end{cases}$$

4. Verify that $f$ is a Valid PDFTo be a valid PDF, the function must be non-negative and integrate to 1 over its domain.Non-negativity: Since $x \in (0,1)$, $x > 0$ and $(1-x) > 0$. The Beta function $B(2,18)$ is positive, so $f(x) \ge 0$.Integration to 1:$$\int_0^1 f(x) \, dx = \int_0^1 \frac{1}{B(2,18)} x^{2-1} (1-x)^{18-1} \, dx = \frac{1}{B(2,18)} \int_0^1 x^{2-1} (1-x)^{18-1} \, dx$$By the definition of the Beta function, $\int_0^1 x^{\alpha-1} (1-x)^{\beta-1} \, dx = B(\alpha, \beta)$. Thus:$$\frac{B(2, 18)}{B(2, 18)} = 1$$

5. Compute the Expectation $E[X]$For a Beta distribution, $E[X] = \frac{\alpha}{\alpha + \beta}$:$$E[X] = \frac{2}{2 + 18} = \frac{2}{20} = \mathbf{0.10}$$

6. Compute the Variance $\text{Var}(X)$The formula for variance is $\text{Var}(X) = \frac{\alpha\beta}{(\alpha+\beta)^2(\alpha+\beta+1)}$:$$\text{Var}(X) = \frac{2 \cdot 18}{(2+18)^2(2+18+1)} = \frac{36}{(20^2)(21)} = \frac{36}{400 \cdot 21} = \frac{36}{8400} = \mathbf{0.004286}$$

7. Probability that Defective Proportion is at most 0.10$$P(X \le 0.10) = \int_0^{0.10} \frac{1}{B(2,18)} x(1-x)^{17} \, dx$$

8. Probability between 0.05 and 0.15$$P(0.05 \le X \le 0.15) = \int_{0.05}^{0.15} \frac{1}{B(2,18)} x(1-x)^{17} \, dx$$

Why a Beta Distribution is ReasonableThe Beta distribution is an ideal model for this situation because:Bounded Domain: It is naturally defined on the interval $[0, 1]$, which matches the physical constraints of a "proportion" or "percentage."Flexibility: By adjusting $\alpha$ and $\beta$, it can represent various shapes of uncertainty, from symmetric to highly skewed (as seen here, where defects are likely low but occasionally higher).

Probability Density Function (PDF) $f_X(x)$The PDF for a Triangular distribution is defined piecewise:$$f_X(x) = \begin{cases} \frac{2(x-a)}{(b-a)(c-a)} & \text{for } a \le x \le c \\ \frac{2(b-x)}{(b-a)(b-c)} & \text{for } c < x \le b \\ 0 & \text{otherwise} \end{cases}$$Plugging in $a=4$, $b=10$, and $c=6$:For $4 \le x \le 6$: $\frac{2(x-4)}{(10-4)(6-4)} = \frac{2(x-4)}{12} = \mathbf{\frac{x-4}{6}}$For $6 < x \le 10$: $\frac{2(10-x)}{(10-4)(10-6)} = \frac{2(10-x)}{24} = \mathbf{\frac{10-x}{12}}$
