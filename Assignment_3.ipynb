Q1) Prediction of Fraudulent Transactions: Answer.


1.Probability Space $(\Omega, \mathcal{F}, P)$Sample Space ($\Omega$):


$$\Omega = \{\omega_1, \omega_2, \omega_3, \omega_4\}$$


$\sigma$-algebra ($\mathcal{F}$): Since

$\mathcal{F} = \mathcal{P}(\Omega)$ (the power set), it contains all $2^4 = 16$ possible subsets of $\Omega$:

$$\mathcal{F} = \{ \emptyset, \{\omega_1\}, \{\omega_2\}, \{\omega_3\}, \{\omega_4\}, \{\omega_1, \omega_2\}, \{\omega_1, \omega_3\}, \{\omega_1, \omega_4\}, \{\omega_2, \omega_3\}, \{\omega_2, \omega_4\}, \{\omega_3, \omega_4\}, \{\omega_1, \omega_2, \omega_3\}, \{\omega_1, \omega_2, \omega_4\}, \{\omega_1, \omega_3, \omega_4\}, \{\omega_2, \omega_3, \omega_4\}, \Omega \}$$

Probability Measure ($P$): Defined on the atoms as:


$$P(\{\omega_1\}) = 0.50, \quad P(\{\omega_2\}) = 0.10, \quad P(\{\omega_3\}) = 0.20, \quad P(\{\omega_4\}) = 0.20$$

For any set $A \in \mathcal{F}$, $P(A) = \sum_{\omega \in A} P(\{\omega\})$.

2. Compute the $\sigma$-algebra Generated by $Y$, $\sigma(Y)$The random variable $Y$ groups the outcomes into two blocks based on customer risk:$Y = 0$ for $\{\omega_1, \omega_2\}$$Y = 1$ for $\{\omega_3, \omega_4\}$The $\sigma$-algebra generated by $Y$ contains all preimages of Borel sets under $Y$:$$\sigma(Y) = \{ \emptyset, \{\omega_1, \omega_2\}, \{\omega_3, \omega_4\}, \Omega \}$$

3. Measurability of $\mathbb{E}[X \mid Y]$By definition, the conditional expectation $\mathbb{E}[X \mid Y]$ is a random variable that must be $\sigma(Y)$-measurable.Because the smallest non-empty sets (atoms) in $\sigma(Y)$ are $A_0 = \{\omega_1, \omega_2\}$ and $A_1 = \{\omega_3, \omega_4\}$, any function that is measurable with respect to $\sigma(Y)$ cannot distinguish between elements within the same atom. Therefore, $\mathbb{E}[X \mid Y]$ must be constant on the sets $\{\omega_1, \omega_2\}$ and $\{\omega_3, \omega_4\}$.

4. Compute $\mathbb{E}[X \mid Y]$ ExplicitlyWe evaluate the expected value of $X$ on each atom of $Y$:For $Y = 0$ (i.e., $\omega \in \{\omega_1, \omega_2\}$):$$\mathbb{E}[X \mid Y=0] = \frac{X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\})}{P(\{\omega_1\}) + P(\{\omega_2\})} = \frac{0(0.50) + 1(0.10)}{0.50 + 0.10} = \frac{0.10}{0.60} = \frac{1}{6}$$For $Y = 1$ (i.e., $\omega \in \{\omega_3, \omega_4\}$):$$\mathbb{E}[X \mid Y=1] = \frac{X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\})}{P(\{\omega_3\}) + P(\{\omega_4\})} = \frac{0(0.20) + 1(0.20)}{0.20 + 0.20} = \frac{0.20}{0.40} = \frac{1}{2}$$Combining these using indicator functions yields:$$\mathbb{E}[X \mid Y] = \frac{1}{6}\mathbb{1}_{\{Y=0\}} + \frac{1}{2}\mathbb{1}_{\{Y=1\}}$$

5. Verify the Defining Property of Conditional ExpectationWe check if $\int_A \mathbb{E}[X \mid Y] dP = \int_A X dP$ holds for all $A \in \sigma(Y)$:For $A = \emptyset$: Both sides are trivially $0$.For $A = \{\omega_1, \omega_2\}$:$$\int_{A} \mathbb{E}[X \mid Y] dP = \frac{1}{6}P(\{\omega_1\}) + \frac{1}{6}P(\{\omega_2\}) = \frac{1}{6}(0.50 + 0.10) = 0.10$$$$\int_{A} X dP = X(\omega_1)P(\{\omega_1\}) + X(\omega_2)P(\{\omega_2\}) = 0(0.50) + 1(0.10) = 0.10$$For $A = \{\omega_3, \omega_4\}$:$$\int_{A} \mathbb{E}[X \mid Y] dP = \frac{1}{2}P(\{\omega_3\}) + \frac{1}{2}P(\{\omega_4\}) = \frac{1}{2}(0.20 + 0.20) = 0.20$$$$\int_{A} X dP = X(\omega_3)P(\{\omega_3\}) + X(\omega_4)P(\{\omega_4\}) = 0(0.20) + 1(0.20) = 0.20$$For $A = \Omega$: Summing both pieces gives $0.10 + 0.20 = 0.30$ on both sides.The defining property is verified.

6. Find the Conditional PMF $p_{X \mid Y}(x \mid y)$Using $p_{X \mid Y}(x \mid y) = \frac{P(X=x, Y=y)}{P(Y=y)}$:For $y = 0$:$$p_{X \mid Y}(0 \mid 0) = \frac{P(\{\omega_1\})}{P(\{\omega_1, \omega_2\})} = \frac{0.50}{0.60} = \frac{5}{6}$$$$p_{X \mid Y}(1 \mid 0) = \frac{P(\{\omega_2\})}{P(\{\omega_1, \omega_2\})} = \frac{0.10}{0.60} = \frac{1}{6}$$For $y = 1$:$$p_{X \mid Y}(0 \mid 1) = \frac{P(\{\omega_3\})}{P(\{\omega_3, \omega_4\})} = \frac{0.20}{0.40} = \frac{1}{2}$$$$p_{X \mid Y}(1 \mid 1) = \frac{P(\{\omega_4\})}{P(\{\omega_3, \omega_4\})} = \frac{0.20}{0.40} = \frac{1}{2}$$

7. Verify the Identity $\mathbb{E}[X \mid Y=y] = \sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid y)$For $y = 0$:$$\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 0) = 0\left(\frac{5}{6}\right) + 1\left(\frac{1}{6}\right) = \frac{1}{6} = \mathbb{E}[X \mid Y=0]$$For $y = 1$:$$\sum_{x \in \{0,1\}} x \cdot p_{X \mid Y}(x \mid 1) = 0\left(\frac{1}{2}\right) + 1\left(\frac{1}{2}\right) = \frac{1}{2} = \mathbb{E}[X \mid Y=1]$$

8. Calculate the Mean Square Error (MSE)The MSE when using the optimal estimator $\mathbb{E}[X \mid Y]$ is given by $\mathbb{E}[(X - \mathbb{E}[X \mid Y])^2]$:$$\text{MSE} = \sum_{i=1}^4 (X(\omega_i) - \mathbb{E}[X \mid Y](\omega_i))^2 \cdot P(\{\omega_i\})$$$$\text{MSE} = \left(0 - \frac{1}{6}\right)^2(0.50) + \left(1 - \frac{1}{6}\right)^2(0.10) + \left(0 - \frac{1}{2}\right)^2(0.20) + \left(1 - \frac{1}{2}\right)^2(0.20)$$$$\text{MSE} = \frac{1}{36}(0.50) + \frac{25}{36}(0.10) + \frac{1}{4}(0.20) + \frac{1}{4}(0.20)$$$$\text{MSE} = \frac{0.5 + 2.5}{36} + \frac{0.4}{4} = \frac{3}{36} + 0.1 = \frac{1}{12} + \frac{1}{10} = \frac{11}{60} \approx 0.1833$$

9. Ignoring $Y$ and the Law of Total VarianceIf we ignore $Y$ and estimate fraud using the global baseline $\mathbb{E}[X] = 0.30$, our estimator becomes a constant. The resulting MSE is simply the total variance of $X$:$$\text{MSE}_{\text{constant}} = \text{Var}(X) = p(1-p) = 0.30 \times 0.70 = 0.21$$Relationship to the Law of Total Variance:The Law of Total Variance states:$$\text{Var}(X) = \mathbb{E}[\text{Var}(X \mid Y)] + \text{Var}(\mathbb{E}[X \mid Y])$$Where:$\mathbb{E}[\text{Var}(X \mid Y)]$ is the minimized MSE achieved with knowledge of $Y$ ($\frac{11}{60} \approx 0.1833$).$\text{Var}(\mathbb{E}[X \mid Y])$ represents the variance captured or explained by knowing $Y$:$$\text{Var}(\mathbb{E}[X \mid Y]) = 0.60\left(\frac{1}{6} - 0.3\right)^2 + 0.40\left(\frac{1}{2} - 0.3\right)^2 = \frac{2}{75} \approx 0.0267$$Verification:$$0.1833 + 0.0267 = 0.21 = \text{Var}(X)$$Ignoring $Y$ increases your error by exactly $\text{Var}(\mathbb{E}[X \mid Y])$.

10. Information Theory Analysisa) Entropy $H(X)$:$$H(X) = -0.3 \log_2(0.3) - 0.7 \log_2(0.7) \approx 0.8813 \text{ bits}$$b) Conditional Entropy $H(X \mid Y)$:$$H(X \mid Y = 0) = -\frac{5}{6}\log_2\left(\frac{5}{6}\right) - \frac{1}{6}\log_2\left(\frac{1}{6}\right) \approx 0.6500 \text{ bits}$$$$H(X \mid Y = 1) = -\frac{1}{2}\log_2\left(\frac{1}{2}\right) - \frac{1}{2}\log_2\left(\frac{1}{2}\right) = 1.0000 \text{ bit}$$$$H(X \mid Y) = P(Y=0)H(X \mid Y=0) + P(Y=1)H(X \mid Y=1)$$$$H(X \mid Y) = 0.60(0.6500) + 0.40(1.0000) = 0.3900 + 0.40 = 0.7900 \text{ bits}$$c) Mutual Information $I(X; Y)$:$$I(X; Y) = H(X) - H(X \mid Y) = 0.8813 - 0.7900 = 0.0913 \text{ bits}$$d) Uncertainty Reduction Percentage:$$\text{Percentage Removed} = \frac{I(X;Y)}{H(X)} \times 100\% = \frac{0.0913}{0.8813} \times 100\% \approx 10.36\%$$

```python
import numpy as np

# Probabilities of the sample space outcomes
P = np.array([0.50, 0.10, 0.20, 0.20])

# Random Variables mappings for omega_1, omega_2, omega_3, omega_4
X = np.array([0, 1, 0, 1])
Y = np.array([0, 0, 1, 1])

# 4. Conditional Expectation E[X|Y]
E_X_given_Y0 = np.sum(X[Y==0] * P[Y==0]) / np.sum(P[Y==0])
E_X_given_Y1 = np.sum(X[Y==1] * P[Y==1]) / np.sum(P[Y==1])
E_X_Y = np.where(Y == 0, E_X_given_Y0, E_X_given_Y1)

print(f"E[X|Y=0] = {E_X_given_Y0:.4f} (Expected: 1/6 ≈ 0.1667)")
print(f"E[X|Y=1] = {E_X_given_Y1:.4f} (Expected: 1/2 = 0.5000)\n")

# 8. MSE Calculation
mse_with_Y = np.sum(((X - E_X_Y) ** 2) * P)
print(f"8) MSE with knowledge of Y: {mse_with_Y:.4f} (Expected: 11/60 ≈ 0.1833)")

# 9. Global estimate baseline (Ignoring Y)
E_X = np.sum(X * P)
mse_without_Y = np.sum(((X - E_X) ** 2) * P)
var_E_X_Y = np.sum(((E_X_Y - E_X) ** 2) * P)

print(f"9) MSE without Y (Var(X)): {mse_without_Y:.4f}")
print(f"   Variance explained by Y (Var(E[X|Y])): {var_E_X_Y:.4f}")
print(f"   Check Law of Total Variance (MSE + Explained): {mse_with_Y + var_E_X_Y:.4f}\n")

# 10. Information Theory computations
def entropy(probs):
    probs = probs[probs > 0]
    return -np.sum(probs * np.log2(probs))

H_X = entropy(np.array([np.sum(P[X==0]), np.sum(P[X==1])]))

# Conditional entropies
P_Y0, P_Y1 = np.sum(P[Y==0]), np.sum(P[Y==1])
H_X_Y0 = entropy(P[Y==0] / P_Y0)
H_X_Y1 = entropy(P[Y==1] / P_Y1)
H_X_given_Y = P_Y0 * H_X_Y0 + P_Y1 * H_X_Y1

I_XY = H_X - H_X_given_Y
pct_reduction = (I_XY / H_X) * 100

print(f"10a) H(X) = {H_X:.4f} bits")
print(f"10b) H(X|Y) = {H_X_given_Y:.4f} bits")
print(f"10c) I(X;Y) = {I_XY:.4f} bits")
print(f"10d) Percentage of uncertainty removed: {pct_reduction:.2f}%")
```

Q2) Conditional Density of the Multivariate Gaussian: Answer


Proof: Conditional Density of the Multivariate GaussianTo find the conditional distribution $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y})$, we use the definition of conditional probability for continuous random variables:$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y}) = \frac{f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})}{f_{\mathbf{Y}}(\mathbf{y})}$$Because $f_{\mathbf{Y}}(\mathbf{y})$ does not depend on $\mathbf{x}$, the conditional density $f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y})$ is proportional to the joint density $f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})$ with respect to $\mathbf{x}$. By expanding the joint density and separating the terms involving $\mathbf{x}$ and $\mathbf{y}$, we can isolate the exact conditional distribution.

1. Partitioning the Precision MatrixLet the precision matrix (the inverse of the covariance matrix) be defined as $\mathbf{\Lambda} = \mathbf{\Sigma}^{-1}$. Using block matrix inversion on partitioned matrices, we can write:$$\mathbf{\Lambda} = \begin{bmatrix} \mathbf{\Lambda}_{XX} & \mathbf{\Lambda}_{XY} \\ \mathbf{\Lambda}_{YX} & \mathbf{\Lambda}_{YY} \end{bmatrix}$$From the properties of Schur complements in block inversion, the top-left block $\mathbf{\Lambda}_{XX}$ and the off-diagonal block $\mathbf{\Lambda}_{XY}$ are given by:$\mathbf{\Lambda}_{XX} = (\mathbf{\Sigma}_{XX} - \mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1}\mathbf{\Sigma}_{YX})^{-1} = \mathbf{\Sigma}_{X|Y}^{-1}$$\mathbf{\Lambda}_{XY} = -\mathbf{\Lambda}_{XX}\mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1} = -\mathbf{\Sigma}_{X|Y}^{-1}\mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1}$

2. Expanding the ExponentThe exponent of the joint distribution contains the quadratic form $Q(\mathbf{x}, \mathbf{y})$:$$Q(\mathbf{x}, \mathbf{y}) = \begin{bmatrix} \mathbf{x} - \boldsymbol{\mu}_X \\ \mathbf{y} - \boldsymbol{\mu}_Y \end{bmatrix}^T \begin{bmatrix} \mathbf{\Lambda}_{XX} & \mathbf{\Lambda}_{XY} \\ \mathbf{\Lambda}_{YX} & \mathbf{\Lambda}_{YY} \end{bmatrix} \begin{bmatrix} \mathbf{x} - \boldsymbol{\mu}_X \\ \mathbf{y} - \boldsymbol{\mu}_Y \end{bmatrix}$$Expanding this product yields:$$Q(\mathbf{x}, \mathbf{y}) = (\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X) + 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XY} (\mathbf{y} - \boldsymbol{\mu}_Y) + (\mathbf{y} - \boldsymbol{\mu}_Y)^T \mathbf{\Lambda}_{YY} (\mathbf{y} - \boldsymbol{\mu}_Y)$$Substitute $\mathbf{\Lambda}_{XY} = -\mathbf{\Lambda}_{XX}\mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1}$ into the cross-term:$$2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XY} (\mathbf{y} - \boldsymbol{\mu}_Y) = -2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} \mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y)$$Now, let's look at the terms depending on $\mathbf{x}$:$$Q_{\mathbf{x}} = (\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X) - 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} \left[ \mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1} (\mathbf{y} - \boldsymbol{\mu}_Y) \right]$$

3. Completing the SquareLet $\mathbf{v} = \mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1}(\mathbf{y} - \boldsymbol{\mu}_Y)$. Notice that the conditional mean provided in the problem is $\boldsymbol{\mu}^* = \boldsymbol{\mu}_X + \mathbf{v}$.Substitute $\mathbf{v}$ into our $Q_{\mathbf{x}}$ equation and complete the square for $\mathbf{x}$:$$Q_{\mathbf{x}} = (\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X) - 2(\mathbf{x} - \boldsymbol{\mu}_X)^T \mathbf{\Lambda}_{XX} \mathbf{v}$$$$Q_{\mathbf{x}} = (\mathbf{x} - \boldsymbol{\mu}_X - \mathbf{v})^T \mathbf{\Lambda}_{XX} (\mathbf{x} - \boldsymbol{\mu}_X - \mathbf{v}) - \mathbf{v}^T \mathbf{\Lambda}_{XX} \mathbf{v}$$Because $\mathbf{x} - \boldsymbol{\mu}_X - \mathbf{v} = \mathbf{x} - (\boldsymbol{\mu}_X + \mathbf{v}) = \mathbf{x} - \boldsymbol{\mu}^*$ and $\mathbf{\Lambda}_{XX} = \mathbf{\Sigma}_{X|Y}^{-1}$, the first term becomes:$$(\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)$$This precisely matches the exponent of the target conditional distribution.(Note: The remaining terms $-\mathbf{v}^T \mathbf{\Lambda}_{XX} \mathbf{v}$ and $(\mathbf{y} - \boldsymbol{\mu}_Y)^T \mathbf{\Lambda}_{YY} (\mathbf{y} - \boldsymbol{\mu}_Y)$ depend only on $\mathbf{y}$ and flawlessly collapse into the exponent of the marginal distribution $f_{\mathbf{Y}}(\mathbf{y})$).

4. Factorizing the Determinant (Normalization Constant)Using the determinant property of block matrices and the Schur complement, the determinant of the full covariance matrix $\mathbf{\Sigma}$ factors as:$$|\mathbf{\Sigma}| = |\mathbf{\Sigma}_{YY}| \cdot |\mathbf{\Sigma}_{XX} - \mathbf{\Sigma}_{XY}\mathbf{\Sigma}_{YY}^{-1}\mathbf{\Sigma}_{YX}| = |\mathbf{\Sigma}_{YY}| \cdot |\mathbf{\Sigma}_{X|Y}|$$Therefore, the normalization constant of the joint distribution easily splits:$$\frac{1}{\sqrt{(2\pi)^{n+m}|\mathbf{\Sigma}|}} = \left( \frac{1}{\sqrt{(2\pi)^n|\mathbf{\Sigma}_{X|Y}|}} \right) \left( \frac{1}{\sqrt{(2\pi)^m|\mathbf{\Sigma}_{YY}|}} \right)$$

5. Final ConclusionBy combining the factored determinant and the factored exponent, the joint density $f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y})$ fully decomposes into two valid probability density functions:$$f_{\mathbf{X},\mathbf{Y}}(\mathbf{x},\mathbf{y}) = \left[ \frac{1}{\sqrt{(2\pi)^n|\mathbf{\Sigma}_{X|Y}|}} \exp\left(-\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)\right) \right] \cdot f_{\mathbf{Y}}(\mathbf{y})$$Dividing both sides by $f_{\mathbf{Y}}(\mathbf{y})$ proves the conditional density relationship:$$f_{\mathbf{X}|\mathbf{Y}}(\mathbf{x}|\mathbf{y}) = \frac{1}{\sqrt{(2\pi)^n|\mathbf{\Sigma}_{X|Y}|}} \exp\left(-\frac{1}{2}(\mathbf{x} - \boldsymbol{\mu}^*)^T \mathbf{\Sigma}_{X|Y}^{-1} (\mathbf{x} - \boldsymbol{\mu}^*)\right)$$

Q3) Analysis: Continuous Fault Monitoring of a Spindle Motor

1. Properties of the Given VariablesFrom the problem statement, we have two independent random variables:Hidden Temperature ($X$): $X \sim \mathcal{N}(0, 1)$Mean: $\mu_X = 0$Variance: $\text{Var}(X) = \Sigma_{XX} = 1$Mechanical Load ($Q$): $Q \sim \mathcal{N}(0, 4)$Mean: $\mu_Q = 0$Variance: $\text{Var}(Q) = 4$$X$ and $Q$ are independent.

2. Deriving the Distribution of the Measurement ($Y$)The measurement is given by the linear combination $Y = X + Q$. Because $Y$ is a sum of independent normal variables, it is also normally distributed.Mean of $Y$: $$\mu_Y = E[X + Q] = E[X] + E[Q] = 0 + 0 = 0Variance of $Y$: $$\text{Var}(Y) = \Sigma_{YY} = \text{Var}(X + Q) = \text{Var}(X) + \text{Var}(Q) = 1 + 4 = 5Thus, the marginal distribution of the measurement is $Y \sim \mathcal{N}(0, 5)$.

3. Finding the Covariance and Joint DistributionTo form the joint distribution, we need the covariance between $X$ and $Y$:$$\text{Cov}(X, Y) = \Sigma_{XY} = \text{Cov}(X, X + Q) = \text{Cov}(X, X) + \text{Cov}(X, Q)$$Because $X$ and $Q$ are independent, their covariance is zero ($\text{Cov}(X, Q) = 0$). Therefore:$$\Sigma_{XY} = \text{Var}(X) = 1$$By symmetry, $\Sigma_{YX} = 1$.The Joint Multivariate Normal Distribution of the vector $\begin{bmatrix} X \\ Y \end{bmatrix}$ is defined by its mean vector and covariance matrix:$$\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathcal{N} \left( \begin{bmatrix} 0 \\ 0 \end{bmatrix}, \begin{bmatrix} 1 & 1 \\ 1 & 5 \end{bmatrix} \right)$$

4. The Conditional Distribution (Estimating Hidden Temperature)Using the formulas established in Q2, we can find the posterior distribution of the hidden temperature $X$ given a specific observed power measurement $Y = y$.Conditional Mean ($\mu_{X|Y}$):$$  \mu_{X|Y} = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y - \mu_Y)$$$$  \mu_{X|Y} = 0 + 1 \cdot (5)^{-1} \cdot (y - 0) = \frac{1}{5}y = 0.2y$$Conditional Variance ($\Sigma_{X|Y}$):$$  \Sigma_{X|Y} = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$$$$  \Sigma_{X|Y} = 1 - 1 \cdot \left(\frac{1}{5}\right) \cdot 1 = 1 - \frac{1}{5} = \frac{4}{5} = 0.8$$Conclusion:Given an observed total spindle power $y$, the estimated hidden internal temperature follows the conditional distribution:$$f_{X|Y}(x|y) \sim \mathcal{N}(0.2y, 0.8)$$
