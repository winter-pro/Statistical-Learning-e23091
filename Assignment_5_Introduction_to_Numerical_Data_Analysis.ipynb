# Section 1: Foundations of Statistical Inference & Hypothesis Testing

### Part A: Theoretical Fundamentals (Maximum Likelihood & Decision Space)

#### 1. Formally derive why $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ is inherently biased in finite samples:
Let $\mathbf{X}_1, \mathbf{X}_2, \dots, \mathbf{X}_n \in \mathbb{R}^m$ be i.i.d. random vectors sampled from a distribution with population mean $\boldsymbol{\mu}$ and population covariance matrix $\boldsymbol{\Sigma}$. The sample mean is defined as:
$$\widehat{\boldsymbol{\mu}}_n = \frac{1}{n}\sum_{i=1}^n \mathbf{X}_i$$

The raw Maximum Likelihood Estimator (MLE) for the covariance matrix is:
$$\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n}\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

We can use an algebraic identity to center the terms around the true population mean $\boldsymbol{\mu}$ instead of the empirical sample mean $\widehat{\boldsymbol{\mu}}_n$:
$$\sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T = \sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T + n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Rearranging terms to isolate the sum of squared errors from the sample mean:
$$\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T = \sum_{i=1}^n (\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T - n(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T$$

Taking the mathematical expectation on both sides:
$$\mathbb{E}\left[\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = \sum_{i=1}^n \mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] - n\,\mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T]$$

Because the vectors $\mathbf{X}_i$ are i.i.d., we know by definition that $\mathbb{E}[(\mathbf{X}_i - \boldsymbol{\mu})(\mathbf{X}_i - \boldsymbol{\mu})^T] = \boldsymbol{\Sigma}$.
Furthermore, the covariance of the sample mean is $\text{Var}(\widehat{\boldsymbol{\mu}}_n) = \mathbb{E}[(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})^T] = \frac{1}{n}\boldsymbol{\Sigma}$.

Substituting these back into our expectation equation yields:
$$\mathbb{E}\left[\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = n\boldsymbol{\Sigma} - n\left(\frac{1}{n}\boldsymbol{\Sigma}\right) = (n-1)\boldsymbol{\Sigma}$$

By multiplying across by the scaling factor $\frac{1}{n}$ from the MLE definition, we find:
$$\mathbb{E}[\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}] = \frac{1}{n} (n-1)\boldsymbol{\Sigma} = \frac{n-1}{n}\boldsymbol{\Sigma}$$
This explicitly demonstrates that $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ is systematically biased, underestimating the true population covariance in finite samples.

#### Demonstration of Bessel's Correction:
To construct an unbiased estimator $\mathbf{S}$, we require $\mathbb{E}[\mathbf{S}] = \boldsymbol{\Sigma}$. By scaling $\widehat{\boldsymbol{\Sigma}}_{\text{MLE}}$ by the inverse ratio of its bias, we apply Bessel's Correction:
$$\mathbf{S} = \frac{n}{n-1}\widehat{\boldsymbol{\Sigma}}_{\text{MLE}} = \frac{1}{n-1}\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T$$

Evaluating the expectation confirms it is unbiased:
$$\mathbb{E}[\mathbf{S}] = \frac{1}{n-1} \mathbb{E}\left[\sum_{i=1}^n (\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)(\mathbf{X}_i - \widehat{\boldsymbol{\mu}}_n)^T\right] = \frac{n-1}{n-1}\boldsymbol{\Sigma} = \boldsymbol{\Sigma}$$

#### 2. Statistical Errors & Confidence Ellipsoids in Continuous Diagnostics
* **Type I Error ($\alpha$):** Declaring a healthy structural asset as anomalous or damaged (a false alarm). In practice, this leads to unnecessary plant shutdowns and costly inspection diagnostics for an asset operating normally.
* **Type II Error ($\beta$):** Failing to flag a structural asset that has experienced true structural degradation or damage (a missed detection). In practice, this risks catastrophic, unmonitored baseline structural failure.
* **Consequence of an ultra-conservative significance threshold ($\alpha = 0.0001$):** Setting a tiny $\alpha$ boundary mathematically forces the critical threshold value of the test statistic outward. This dramatically **expands the geometric volume of the "healthy operation" confidence ellipsoid**, making the model exceptionally lenient. While false alarms are virtually eliminated, the mathematical consequence is a massive increase in the probability of a Type II error ($\beta$). This severely drives down the **statistical power ($1 - \beta$)** of the diagnostic layer, rendering it blind to low-magnitude or early-stage structural changes.

### Part B: Theoretical Extension (Asymptotic Distributions & Slutsky's Theorem)

#### 1. Rigorous Statement of Slutsky's Theorem
Let $X_n$ be a sequence of random vectors that converges in distribution to a random variable $X$ ($X_n \xrightarrow{d} X$), and let $Y_n$ be a sequence of random matrices that converges in probability to a constant matrix $\mathbf{C}$ ($Y_n \xrightarrow{p} \mathbf{C}$). Then:
1. $X_n + Y_n \xrightarrow{d} X + \mathbf{C}$
2. $Y_n X_n \xrightarrow{d} \mathbf{C} X$
3. $Y_n^{-1} X_n \xrightarrow{d} \mathbf{C}^{-1} X$ (provided $\mathbf{C}$ is nonsingular and invertible).

#### 2. Justification for Substituting the Empirical Matrix $\mathbf{S}$
By the Weak Law of Large Numbers (WLLN), the empirical sample covariance matrix is a consistent estimator of the population matrix, converging in probability as the snapshot sample size grows:
$$\mathbf{S} \xrightarrow{p} \boldsymbol{\Sigma} \quad \text{as } n \to \infty$$
By the Continuous Mapping Theorem, the inverse square root transformation maintains this property:
$$\mathbf{S}^{-1/2} \xrightarrow{p} \boldsymbol{\Sigma}^{-1/2}$$

The Lindeberg-Lévy Multivariate Central Limit Theorem (CLT) states that:
$$\sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \xrightarrow{d} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma})$$

We define our standardized mapping sequence as $X_n = \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu})$ and the multiplier matrix sequence as $Y_n = \mathbf{S}^{-1/2}$. Applying Slutsky's Theorem:
$$\mathbf{S}^{-1/2} \left[ \sqrt{n}(\widehat{\boldsymbol{\mu}}_n - \boldsymbol{\mu}) \right] \xrightarrow{d} \boldsymbol{\Sigma}^{-1/2} \mathscr{N}(\mathbf{0}, \boldsymbol{\Sigma}) \sim \mathscr{N}(\mathbf{0}, \mathbf{I})$$

Because the limit distribution is structurally identical whether scaled by the unobservable parameter $\boldsymbol{\Sigma}$ or the empirical matrix $\mathbf{S}$, we can scale the identity matrix back out asymptotically to justify the practical parametric distribution:
$$\widehat{\boldsymbol{\mu}}_n \sim \mathscr{N}\left(\boldsymbol{\mu}, \frac{1}{n}\mathbf{S}\right)$$

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

# --- DO NOT MODIFY: Synthetic Strain Sensor Data Generator ---
np.random.seed(42)
n_samples = 5000
n_features = 3

base_data = np.random.multivariate_normal(
    mean=[0.5, -0.2, 1.1],
    cov=[[0.09, 0.02, 0.01],
         [0.02, 0.06, 0.03],
         [0.01, 0.03, 0.05]],
    size=n_samples
)

# Injecting a subtle, hidden structural drift in the final 1,000 snapshots
base_data[4000:, 0] += 0.015  # Sensor 1 drift
base_data[4000:, 2] -= 0.010  # Sensor 3 drift

df_strain = pd.DataFrame(base_data, columns=['Strain_Ch1', 'Strain_Ch2', 'Strain_Ch3'])
# ----------------------------------------------------------------

def verify_first_moment_homogeneity(df: pd.DataFrame, g_chunks: int = 5) -> dict:
    """
    Partitions the dataset into g chunks and evaluates first-moment homogeneity
    via Wilks' Lambda and Bartlett's Chi-Square asymptotic transformation.
    """
    n = len(df)
    m = df.shape[1]
    chunk_size = n // g_chunks

    # Compute global mean
    global_mean = df.mean().values

    W = np.zeros((m, m))
    B = np.zeros((m, m))

    # Calculate Within-Chunk and Between-Chunk Matrices
    for g in range(g_chunks):
        chunk = df.iloc[g * chunk_size : (g + 1) * chunk_size]
        chunk_mean = chunk.mean().values

        # Within matrix contribution
        diff_within = chunk.values - chunk_mean
        W += diff_within.T @ diff_within

        # Between matrix contribution
        diff_between = chunk_mean - global_mean
        B += chunk_size * np.outer(diff_between, diff_between)

    # Wilks' Lambda calculation
    wilks_lambda = np.linalg.det(W) / np.linalg.det(B + W)

    # Bartlett's Chi-Square Approximation Transformation
    degrees_of_freedom = m * (g_chunks - 1)
    chi2_calc = - (n - 1 - (m + g_chunks) / 2.0) * np.log(wilks_lambda)
    p_value = 1.0 - stats.chi2.cdf(chi2_calc, degrees_of_freedom)

    return {
        "Wilks_Lambda": wilks_lambda,
        "Chi2_Stat": chi2_calc,
        "Degrees_of_Freedom": degrees_of_freedom,
        "p_value": p_value
    }

# Execute completed function
results = verify_first_moment_homogeneity(df_strain, g_chunks=5)
print("--- MANOVA / Bartlett First-Moment Homogeneity Results ---")
for k, v in results.items():
    print(f"{k}: {v:.6e}" if k == "p_value" else f"{k}: {v:.5f}")

# Diagnostic Conclusion Statement
alpha = 0.05
print("\n--- Diagnostic Conclusion ---")
if results["p_value"] < alpha:
    print(f"Reject H0 at alpha={alpha}. Conclusion: The baseline center has shifted over time. The first moment is HETEROGENEOUS.")
else:
    print(f"Fail to Reject H0 at alpha={alpha}. Conclusion: No significant baseline shift detected. The first moment is HOMOGENEOUS.")
```

# Section 2: Geometric Subspace Optimization via PCA

### Part A: Theoretical Fundamentals (Coordinate Projections & Orthogonality)

#### 1. Step-by-step proof that $\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{\Lambda}$:
Given the linear projection definition $\mathbf{Z}_i = \mathbf{P}^T \widetilde{\mathbf{X}}_i$, the theoretical covariance matrix of the transformed variables can be written as:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[(\mathbf{P}^T \widetilde{\mathbf{X}}_i)(\mathbf{P}^T \widetilde{\mathbf{X}}_i)^T\right]$$

Applying the matrix transpose property $(AB)^T = B^T A^T$:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[\mathbf{P}^T \widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T \mathbf{P}\right]$$

Since the projection loading matrix $\mathbf{P}$ contains deterministic constants, we can factor it completely outside of the expectation operator:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \mathbb{E}\left[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T\right] \mathbf{P}$$

By definition, the population covariance of the centered vectors is $\mathbb{E}\left[\widetilde{\mathbf{X}}_i \widetilde{\mathbf{X}}_i^T\right] = \boldsymbol{\Sigma}$:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T \boldsymbol{\Sigma} \mathbf{P}$$

Substituting the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ into the expression:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{P}^T (\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) \mathbf{P}$$

Grouping matrices due to associativity:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = (\mathbf{P}^T \mathbf{P}) \mathbf{\Lambda} (\mathbf{P}^T \mathbf{P})$$

Because $\mathbf{P}$ is an orthogonal matrix, its transpose equals its inverse, meaning $\mathbf{P}^T \mathbf{P} = \mathbf{I}_{m \times m}$. Thus:
$$\mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbf{I} \mathbf{\Lambda} \mathbf{I} = \mathbf{\Lambda}$$

* **Geometric & Statistical Interpretation:** Because $\mathbf{\Lambda}$ is strictly diagonal, the cross-covariances between distinct coordinates $Z_{i,j}$ and $Z_{i,k}$ (where $j \neq k$) are zero. Statistically, this proves that PCA transforms the raw, highly correlated feature entries into a set of completely uncorrelated coordinates. Geometrically, it means the raw features have been projected onto a new coordinate frame aligned with the directions of maximum variance, removing all linear redundancy.

#### 2. Proof of Total System Variance Conservation
Using the spectral decomposition $\boldsymbol{\Sigma} = \mathbf{P} \mathbf{\Lambda} \mathbf{P}^T$ inside the matrix trace:
$$\text{tr}(\boldsymbol{\Sigma}) = \text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T)$$

Applying the cyclic permutation invariance property of the trace matrix operator ($\text{tr}(ABC) = \text{tr}(CAB)$):
$$\text{tr}(\mathbf{P} \mathbf{\Lambda} \mathbf{P}^T) = \text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda})$$

Since $\mathbf{P}$ is orthogonal, $\mathbf{P}^T \mathbf{P} = \mathbf{I}$. This simplifies directly to:
$$\text{tr}(\mathbf{P}^T \mathbf{P} \mathbf{\Lambda}) = \text{tr}(\mathbf{I} \mathbf{\Lambda}) = \text{tr}(\mathbf{\Lambda}) = \sum_{j=1}^m \lambda_j$$
This proves that linear rotations do not alter or destroy total system energy; they merely redistribute variance across orthogonal vectors.

* **Subspace Ratio Formulations:**
  * **Cumulative Explained Variance Ratio $\Phi(k)$:**
    $$\Phi(k) = \frac{\sum_{j=1}^k \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j}$$
  * **Residual Unexplained Variance Ratio $\Psi(k)$:**
    $$\Psi(k) = 1 - \Phi(k) = \frac{\sum_{j=k+1}^m \widehat{\lambda}_j}{\sum_{j=1}^m \widehat{\lambda}_j}$$

#### 3. Subspace Coordinate Decompositions
* **Definitions:**
  * **Reconstructed Vector $\widehat{\mathbf{x}}_i$:** $\widehat{\mathbf{x}}_i = \widehat{\boldsymbol{\mu}}_n + \widehat{\mathbf{P}}_k \mathbf{z}_{i,k}$
  * **Residual Error Vector $\mathbf{e}_i$:** $\mathbf{e}_i = (\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n) - \widehat{\mathbf{x}}_i = \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$

* **Proof via Pythagorean Identity:**
  Let the centered observation be $\widetilde{\mathbf{x}}_i = \mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n = \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$. Its total squared Euclidean norm is:
  $$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \widetilde{\mathbf{x}}_i^T \widetilde{\mathbf{x}}_i = \left(\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\right)^T \left(\widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}\right)$$
  Expanding the multiplication:
  $$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k} + 2\mathbf{z}_{i,k}^T \widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} \mathbf{z}_{i,m-k}$$
  Because the principal axes matrices represent mutually orthogonal subspaces, the cross-subspace inner product matrix is explicitly zero ($\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_{m-k} = \mathbf{0}$). Furthermore, since the individual components are orthonormal, $\widehat{\mathbf{P}}_k^T \widehat{\mathbf{P}}_k = \mathbf{I}$ and $\widehat{\mathbf{P}}_{m-k}^T \widehat{\mathbf{P}}_{m-k} = \mathbf{I}$. This leaves:
  $$\|\mathbf{x}_i - \widehat{\boldsymbol{\mu}}_n\|^2 = \mathbf{z}_{i,k}^T \mathbf{I} \mathbf{z}_{i,k} + \mathbf{z}_{i,m-k}^T \mathbf{I} \mathbf{z}_{i,m-k} = \|\mathbf{z}_{i,k}\|^2 + \|\mathbf{z}_{i,m-k}\|^2$$

* **Diagnostic Contrast ($T^2$ vs $Q$ Statistics):**
  * **Hotelling's $T^2$** monitors variations *within* the nominal principal subspace (plane of expected operating states).
  * **Residual $Q$ (SPE)** monitors variations *orthogonal* to the principal subspace (deviations away from the normal plane).
  * **Matching Operational Events:**
    * **Environmental Load Anomaly (e.g., Extreme Wind):** Maps to **$T^2$**. Strong operational winds drive extreme structural motion, but this motion still respects the asset's standard physical degrees of freedom. The data remains within the defined principal subspace plane but moves far out from the nominal origin.
    * **Internal Structural Fracture (e.g., Fatigue Crack):** Maps to **$Q$**. A physical crack alters the system's structural configuration and changes its underlying load pathways, introducing unmodeled cross-correlations. This breaks the linear dependencies captured during baseline training, causing the observation vector to escape the principal subspace and spiking the residual matrix energy.

```python
class PCASubspaceMonitor:
    def __init__(self, data: pd.DataFrame):
        self.raw_data = data.values
        self.n, self.m = data.shape
        self.means = np.mean(self.raw_data, axis=0)
        self.stds = np.std(self.raw_data, axis=0, ddof=1)

    def execute_analysis(self):
        # 1. Standardize and center
        Z = (self.raw_data - self.means) / self.stds

        # 2. Covariance decomposition
        S = (Z.T @ Z) / (self.n - 1)
        eigenvalues, eigenvectors = np.linalg.eigh(S)

        # Sort in descending order
        idx = np.argsort(eigenvalues)[::-1]
        lambdas = eigenvalues[idx]
        P = eigenvectors[:, idx]

        # Project data to get all principal components
        Z_proj = Z @ P

        metrics = {}
        # 3. Loop over truncation cutoff values k
        for k in [1, 2, 3]:
            # Hotelling's T2 calculation per sample
            t2_samples = np.sum((Z_proj[:, :k] ** 2) / lambdas[:k], axis=1)

            # Residual Q statistic per sample
            # Reconstruct using k components
            Z_hat = Z_proj[:, :k] @ P[:, :k].T
            residuals = Z - Z_hat
            q_samples = np.sum(residuals ** 2, axis=1)

            # 4. Compute empirical means
            metrics[k] = {
                "Expected_T2": np.mean(t2_samples),
                "Expected_Q": np.mean(q_samples)
            }
        return metrics

# Run tracking execution
np.random.seed(42)
mock_baseline = pd.DataFrame(np.random.multivariate_normal([0,0,0,0], np.eye(4), 3000))
pca_pipeline = PCASubspaceMonitor(mock_baseline)
summary_metrics = pca_pipeline.execute_analysis()

print("--- Subspace Truncation Metric Summary ---")
for k, val in summary_metrics.items():
    print(f"Truncation Cutoff k = {k} -> E[T²]: {val['Expected_T2']:.4f} | E[Q]: {val['Expected_Q']:.4f}")
```

# Section 3: Latent Subspace Decomposition via Factor Analysis (FA)

### Part A: Theoretical Fundamentals (Generative Model & Commonalities)

#### 1. Prove the Fundamental Equation of Factor Analysis: $\mathbf{R} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$
Given the linear generative model framework $\mathbf{Z}_i = \boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i$, we evaluate the universal correlation matrix:
$$mathbf{R} = \mathbb{E}[\mathbf{Z}_i \mathbf{Z}_i^T] = \mathbb{E}\left[(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)(\boldsymbol{\Lambda} \mathbf{F}_i + \boldsymbol{\epsilon}_i)^T\right]$$

Expanding the terms inside the expectation operator using matrix transposition:
$$\mathbf{R} = \mathbb{E}\left[\boldsymbol{\Lambda} \mathbf{F}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\epsilon}_i \mathbf{F}_i^T \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbf{F}_i \boldsymbol{\epsilon}_i^T + \boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T\right]$$

Using the linearity of expectations and factoring out deterministic matrices:
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] \boldsymbol{\Lambda}^T + \boldsymbol{\Lambda} \mathbb{E}[\mathbf{F}_i \boldsymbol{\epsilon}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T]$$

Applying the core structural constraints defined by the Factor Analysis paradigm:
1. Common factors are orthogonal and standardized: $\mathbb{E}[\mathbf{F}_i \mathbf{F}_i^T] = \mathbf{I}_{k \times k}$
2. Factor loading states and sensor errors are independent: $\mathbb{E}[\boldsymbol{\epsilon}_i \mathbf{F}_i^T] = \mathbf{0}$
3. Unique errors are mutually uncorrelated: $\mathbb{E}[\boldsymbol{\epsilon}_i \boldsymbol{\epsilon}_i^T] = \boldsymbol{\Psi}$

Substituting these constraints simplifies the expression directly to:
$$\mathbf{R} = \boldsymbol{\Lambda} \mathbf{I} \boldsymbol{\Lambda}^T + \mathbf{0} + \mathbf{0} + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi}$$

* **Physical Interpretations for the $j$-th Diagonal Sensor Entry:**
  The entry is represented by $R_{jj} = 1 = h_j^2 + \varphi_j^2$.
  * **Communality ($h_j^2 = \sum_{r=1}^k \lambda_{jr}^2$):** The portion of a sensor's total variance explained by the shared, global latent physical mechanisms (the common factors). High communality indicates that the sensor is reliable and well-integrated into tracking global system states.
  * **Uniqueness ($\varphi_j^2$):** The portion of a sensor's variance specific only to itself (e.g., localized physical effects or measurement noise). In a continuous diagnostic layer, an elevated uniqueness identifies sensor-specific faults, such as calibration drift or electrical shorts, isolating them from true system-wide changes.

#### 2. Interpretations of Coordinate Rotations
* **Raw PCA Loadings vs. Varimax-Rotated FA Loadings:** Raw PCA loadings focus purely on maximizing variance capture hierarchically along consecutive orthogonal axes. This often yields a complex loading pattern where all sensors have moderate cross-loadings across multiple components. In contrast, Varimax rotates the factor loading axes to maximize the variance of the *squared loadings* within each column. This drives individual loadings toward 1 or 0, yielding a "simple structure" where each sensor loads heavily on exactly one factor, aligning cleanly with specific physical components or subsystems.
* **Why Orthogonal Rotations Do Not Change Communality:** An orthogonal rotation is defined by a transformation matrix $\mathbf{T}$ satisfying $\mathbf{T}\mathbf{T}^T = \mathbf{I}$. The rotated loading matrix is $\boldsymbol{\Lambda}^* = \boldsymbol{\Lambda}\mathbf{T}$. Substituting this into the global correlation matrix expression:
$$\mathbf{R}^* = \boldsymbol{\Lambda}^*(\boldsymbol{\Lambda}^*)^T + \boldsymbol{\Psi} = (\boldsymbol{\Lambda}\mathbf{T})(\boldsymbol{\Lambda}\mathbf{T})^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}(\mathbf{T}\mathbf{T}^T)\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \boldsymbol{\Lambda}\mathbf{I}\boldsymbol{\Lambda}^T + \boldsymbol{\Psi} = \mathbf{R}$$
Because the total reconstructed matrix $\mathbf{R}$ is invariant under orthogonal rotations, its diagonal elements remain identical. Since uniqueness ($\boldsymbol{\Psi}$) is unchanged, the individual sensor communalities ($h_j^2 = R_{jj} - \varphi_j^2$) remain completely invariant.

#### 3. Thomson's Regression Method Score Derivations
* **Derivation of the Partitioned Covariance Matrix $\boldsymbol{\Sigma}_{YY}$:**
  Let the combined column matrix be defined as $\mathbf{Y}_i = [\mathbf{Z}_i^T, \mathbf{F}_i^T]^T$. Its joint covariance matrix structure is:
  $$\boldsymbol{\Sigma}_{YY} = \mathbb{E}[\mathbf{Y}_i \mathbf{Y}_i^T] = \begin{bmatrix} \mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] \\ \mathbb{E}[\mathbf{F}_i\mathbf{Z}_i^T] & \mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] \end{bmatrix}$$

  We know $\mathbb{E}[\mathbf{Z}_i\mathbf{Z}_i^T] = \mathbf{R}$ and $\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] = \mathbf{I}$. Evaluating the cross-covariance component using the generative model equation:
  $$\mathbb{E}[\mathbf{Z}_i\mathbf{F}_i^T] = \mathbb{E}[(\boldsymbol{\Lambda}\mathbf{F}_i + \boldsymbol{\epsilon}_i)\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbb{E}[\mathbf{F}_i\mathbf{F}_i^T] + \mathbb{E}[\boldsymbol{\epsilon}_i\mathbf{F}_i^T] = \boldsymbol{\Lambda}\mathbf{I} + \mathbf{0} = \boldsymbol{\Lambda}$$
  Consequently, $\mathbb{E}[\mathbf{F}_i\mathbf{Z}_i^T] = \boldsymbol{\Lambda}^T$. This constructs the complete joint matrix:
  $$\boldsymbol{\Sigma}_{YY} = \begin{bmatrix} \mathbf{R} & \boldsymbol{\Lambda} \\ \boldsymbol{\Lambda}^T & \mathbf{I}_{k \times k} \end{bmatrix}$$

* **Derivation of Thomson's Factor Score Estimator:**
  Using the standard linear conditional expectation formula for multivariate normal distributions, where the target variable is $\mathbf{F}_i$ given the observation vector $\mathbf{z}_i$:
  $$\mathbf{f}_i = \mathbb{E}[\mathbf{F}_i | \mathbf{z}_i] = \boldsymbol{\mu}_F + \boldsymbol{\Sigma}_{FZ}\boldsymbol{\Sigma}_{ZZ}^{-1}(\mathbf{z}_i - \boldsymbol{\mu}_Z)$$

  Since the centered data vectors have zero means ($\boldsymbol{\mu}_F = \mathbf{0}, \boldsymbol{\mu}_Z = \mathbf{0}$), and substituting the blocks from $\boldsymbol{\Sigma}_{YY}$ ($\boldsymbol{\Sigma}_{ZZ} = \mathbf{R}$ and $\boldsymbol{\Sigma}_{FZ} = \boldsymbol{\Lambda}^T$):
  $$\mathbf{f}_i = \mathbf{0} + \boldsymbol{\Lambda}^T \mathbf{R}^{-1}(\mathbf{z}_i - \mathbf{0}) = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$$

```python
import numpy as np
import pandas as pd
from sklearn.decomposition import FactorAnalysis
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# --- 1. Synthetic Asset Data Generation ---
np.random.seed(42)
n_samples = 2500

# True hidden physical structural dimensions
f1 = np.random.normal(0, 1, n_samples)
f2 = np.random.normal(0, 1, n_samples)

# Generate sensor channels with cross-loadings and an isolated noisy channel (Sensor 4)
s1 = 0.85 * f1 + 0.10 * f2 + np.random.normal(0, 0.3, n_samples)
s2 = 0.80 * f1 + 0.15 * f2 + np.random.normal(0, 0.35, n_samples)
s3 = 0.12 * f1 + 0.90 * f2 + np.random.normal(0, 0.25, n_samples)
s4 = 0.02 * f1 + 0.05 * f2 + np.random.normal(0, 1.40, n_samples)  # Highly unique/noisy

df_asset = pd.DataFrame(
    data=np.vstack([s1, s2, s3, s4]).T,
    columns=['Sensor_1', 'Sensor_2', 'Sensor_3', 'Sensor_4']
)

# Standardize inputs to match universal correlation matrix definitions
Z_asset = (df_asset - df_asset.mean()) / df_asset.std(ddof=1)

# --- 2. Fit Factor Analysis Model with Varimax Rotation ---
fa_engine = FactorAnalysis(n_components=2, rotation='varimax', random_state=42)
fa_engine.fit(Z_asset)

# Extract metrics
loadings = fa_engine.components_.T  # Shape: (m, k)
uniquenesses = fa_engine.noise_variance_
communalities = 1.0 - uniquenesses
sensor_labels = df_asset.columns

# --- 3. Construct Dashboard 1: Subspace Parameter Layout ---
fig_dash1 = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Varimax-Rotated Factor Loadings Matrix", "Variance Composition (Communality vs Uniqueness)"),
    shared_yaxes=True
)

# Left: Loadings Heatmap
fig_dash1.add_trace(
    go.Heatmap(
        z=loadings, x=['Factor 1', 'Factor 2'], y=sensor_labels,
        colorscale='RdBu', zmin=-1, zmax=1, text=np.round(loadings, 3), texttemplate="%{text}",
        colorbar=dict(title="Loading Matrix Weight", x=0.45)
    ), row=1, col=1
)

# Right: Stacked Bar Chart for Communality vs Uniqueness
fig_dash1.add_trace(
    go.Bar(name='Communality (h²)', y=sensor_labels, x=communalities, orientation='h', marker_color='cadetblue'),
    row=1, col=2
)
fig_dash1.add_trace(
    go.Bar(name='Uniqueness (φ²)', y=sensor_labels, x=uniquenesses, orientation='h', marker_color='crimson'),
    row=1, col=2
)

fig_dash1.update_layout(
    title_text="<b>Dashboard Layout 1: Factor Analysis Subspace Decoupling</b>",
    barmode='stack', width=1100, height=500, showlegend=True
)
fig_dash1.show()

# --- 4. Construct Dashboard 2: Operational Subspace Truncation Profiles ---
# Perform a standard PCA loop to generate the comparison elbow curves
S_mat = np.cov(Z_asset.T)
eig_vals, _ = np.linalg.eigh(S_mat)
eig_vals = eig_vals[::-1]

k_axis = [1, 2, 3]
q_profile = [np.sum(eig_vals[k:]) for k in k_axis]
t2_profile = [k for k in k_axis]  # Theoretical standard average dimension size matches k

fig_dash2 = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Residual Noise Floor Optimization (Q Statistic Curve)", "Subspace Tracking Scale (Hotelling's T² Curve)")
)

fig_dash2.add_trace(
    go.Scatter(x=k_axis, y=q_profile, mode='lines+markers', name='Q Noise Floor', line=dict(color='firebrick', width=3)),
    row=1, col=1
)
fig_dash2.add_trace(
    go.Scatter(x=k_axis, y=t2_profile, mode='lines+markers', name='T² Dimension Scale', line=dict(color='navy', width=3)),
    row=1, col=2
)

fig_dash2.update_layout(
    title_text="<b>Dashboard Layout 2: Multi-Scale Diagnostic Optimization Profiles</b>",
    width=1100, height=450, showlegend=False
)
fig_dash2.update_xaxes(title_text="Subspace Truncation Cutoff (k)", tickvals=[1,2,3])
fig_dash2.update_yaxes(title_text="Expected Q Statistic Energy", row=1, col=1)
fig_dash2.update_yaxes(title_text="Expected Subspace Scale Value", row=1, col=2)
fig_dash2.show()
```

### Part B: Operational Diagnostic Interpretations

#### Question 1: The Total Variance Illusion
1. **Uniqueness interpretation for Sensor 4:** The parameter chart reveals Sensor 4's Uniqueness ($\varphi_4^2$) is close to 99%. This proves that Sensor 4 shares virtually none of its variance with the underlying global structural modes ($f_1, f_2$). Its signal consists almost entirely of localized, un-correlated measurement noise.
2. **How localized noise tricks PCA:** PCA treats all forms of variance equally. Because Sensor 4 has a huge noise scale ($\sigma = 1.40$), its raw variance is a dominant component of the overall data matrix. PCA blindly constructs its top principal components to align with this massive noise energy to maximize total variance capture, inflating the leading eigenvalues.
3. **Major Operational Risk:** The risk is that an anomaly detection layer built on unadjusted PCA will be highly sensitive to instrumentation faults and sensor noise on Sensor 4, while remaining blind to or misinterpreting critical shifts in the true shared physical structural dimensions ($f_1, f_2$).

#### Question 2: Decoupling Structural Loading via Rotation
1. **How Varimax rotation creates simple structure:** Standard PCA components mix loadings across all sensor channels to maximize variance hierarchically. Varimax rotates these factor axes to maximize the variance of the *squared loadings* within columns. This drives the weights toward 1 or 0, forcing each sensor to align heavily with exactly one factor and making physical interpretations clear.
2. **Troubleshooting advantage for operators:** If an asset suffers damage that impacts a specific structural loading group, the operator can immediately isolate the root cause because the affected sensors map clearly to a single rotated factor. In unrotated PCA, the anomaly's signature is smeared across multiple components, masking the physical origin.

#### Question 3: Determining Subspace Truncation using T² and Q Profiles
1. **Behavior of the Q profile:** The $Q$ statistic curve drops sharply as truncation changes from $k=1$ to $k=2$. From $k=2$ to $k=3$, the curve flattens out completely into a distinct "flat elbow."
2. **Why the flat elbow identifies $k=2$ as the true dimensionality:** The sharp drop between $k=1$ and $k=2$ shows that the second component captures a massive amount of shared structural variance. The flat elbow after $k=2$ indicates that adding more dimensions captures no new structural info—only random, localized sensor noise.
3. **Consequences of choosing $k=3$:** By choosing $k=3$, you force purely random, localized sensor noise (from Sensor 4) into your "clean" principal subspace. This overfits your normal baseline model, exposing your system to false alarms caused by instrumentation drift.

#### Question 4: Operational Trade-offs in System Health Monitoring
* **Most Robust Strategy:** **The FA (Factor Analysis) Strategy** is significantly more robust against a single sensor losing calibration or shorting out.
* **Justification via Uniqueness Floor ($\boldsymbol{\Psi}$):** In the FA framework, any localized sensor-specific noise or calibration loss is isolated entirely within that sensor's individual uniqueness entry ($\varphi_j^2$) in the diagonal noise matrix $\boldsymbol{\Psi}$. Because factor scores are calculated via $\mathbf{f}_i = \boldsymbol{\Lambda}^T \mathbf{R}^{-1} \mathbf{z}_i$, the model suppresses un-correlated sensor spikes. In contrast, the PCA strategy bakes Sensor 4's massive unique noise directly into its principal eigenvectors. As a result, an instrumentation fault on that sensor will directly corrupt both the Hotelling's $T^2$ and the residual $Q$ statistics, triggering false alarms.
