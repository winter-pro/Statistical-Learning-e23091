## Q1) The Probability of Shooting at a Target

### Model 1: Uniform over area
We assume the point is uniformly distributed with respect to area over the disk $D$ of radius $R=10$.

- Total area $Area(D) = \pi R^2 = 100\pi$.

- Event $A$ is the region within 1 unit of the boundary, i.e., the annulus where $9 \le r \le 10$.

- $Area(A) = \pi(10^2) - \pi(9^2) = 100\pi - 81\pi = 19\pi$.

- $P(A) = \frac{Area(A)}{Area(D)} = \frac{19\pi}{100\pi} = 0.19$.

### Model 2: Radius uniformly distributed
We assume $R \sim Uniform(0,10)$ and direction $\theta \sim Uniform(0, 2\pi)$.

- The probability depends only on $R$.

- $P(A) = P(9 \le R \le 10)$.

- Since $R$ is uniform on $[0,10]$, the PDF is $f_R(r) = 1/10$.

- $P(A) = \int_{9}^{10} \frac{1}{10} dr = \frac{1}{10} = 0.1$.

### Explanation of Difference
The results differ because the "Uniform over area" model implies that the PDF of the radius is $f_R(r) = \frac{2r}{100}$, which places significantly more weight on larger radii (near the boundary) than the uniform radius model ($f_R(r) = 0.1$).





## Q2) Bertrand’s Paradox

We consider an equilateral triangle inscribed in a circle. The side length is $L = \sqrt{3}R$.

1. **Random Endpoints (Method 1):** Choose two points independently and uniformly on the circumference. Rotate the circle so the first point is at a vertex. The chord is longer than $L$ if the second point falls on the arc opposite the vertex, which covers $1/3$ of the circumference. **P = 1/3**.

2. **Random Midpoint (Method 2):** A chord is uniquely determined by its midpoint. A chord is longer than $L$ if its midpoint lies within a concentric circle of radius $R/2$. The ratio of areas is $\frac{\pi(R/2)^2}{\pi R^2} = 1/4$. **P = 1/4**.

3. **Random Radius (Method 3):** Choose a radius and a point on it uniformly. The chord through that point perpendicular to the radius is longer than $L$ if the point is within $R/2$ of the center. **P = 1/2**.

### Conclusion
The paradox demonstrates that "at random" is ill-defined without specifying the probability measure on the space of chords.


## Q8) Probability of Being a Girl

- **Sample Space (Ω):** {BB, BG, GB, GG}, each with probability 1/4.

- **Condition:** At least one child is a girl ($C = \{BG, GB, GG\}$).

- **Event:** Both are girls ($E = \{GG\}$).

- **Calculation:** $P(E|C) = \frac{P(E \cap C)}{P(C)} = \frac{1/4}{3/4} = 1/3$.




## Q9) Information in a Discrete Random Variable

### Probability space
- $\Omega = \{G1, G2, D1, D2\}$

- $F = \mathcal{P}(\Omega)$ (Power set)

- $P(\{G1\})=0.5, P(\{G2\})=0.2, P(\{D1\})=0.1, P(\{D2\})=0.2$

### $\sigma$-algebras
- $\sigma(X) = \{\emptyset, \Omega, \{G1, G2\}, \{D1, D2\}\}$

- $\sigma(Y) = \mathcal{P}(\Omega)$ (Since $Y$ maps each outcome to a unique integer).

- **Inclusion:** $\sigma(X) \subseteq \sigma(Y)$. $Y$ provides more information because knowing $Y$ tells you both the machine and the quality, whereas $X$ only tells you quality.

### Entropy (Calculated in Python below)
- $H(X)$: Uncertainty in quality.

- $H(Y)$: Total uncertainty.

- $H(Y|X)$: Uncertainty about the machine given quality is known.

- $H(X|Y) = 0$: Since $Y$ completely determines $X$.
